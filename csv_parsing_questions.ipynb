Interview Question: Parsing CSV Files
Problem
Football: The football.csv file contains the results from the English Premier League. The columns labeled ‘Goals’ and ‘Goals Allowed’ contain the total number of goals scored for and against each team in that season (so Arsenal scored 79 goals against opponents, and had 36 goals scored against them). Write a program to read the file, then print the name of the team with the smallest difference in ‘for’ and ‘against’ goals.
Weather: In weather.csv you’ll find daily weather data. Write a program to read the file, then output the day number (column one) with the smallest temperature spread (the maximum temperature is the second column, the minimum the third column).
See if you can write the same program to solve both questions.

```python
import os
import csv

min_diff = float('inf')
with open("football.csv") as file:
    rows = csv.DictReader(file)
    for row in rows:
        row["smallest_diff"] = int(row["Goals"]) - int(row["Goals Allowed"])
        if row["smallest_diff"]<min_diff:
          min_diff = row["smallest_diff"]
          min_team = row["Team"]
        print(f"Team Name is {row["Team"]} and difference is {(row["smallest_diff"])}")
    print(min_team)
```

```python
min_spread = float('inf')
with open("weather.csv") as file:
    rows = csv.DictReader(file)
    for row in rows:
        spread = int(row["MxT"]) - int(row["MnT"])
        if spread<min_spread:
            min_spread = spread
            min_day = row["Day"]
    print(min_day)
```

You will be supplied with two data files in CSV format .
The first file contains statistics about various dinosaurs. The second file contains additional data.
Given the following formula, speed = ((STRIDE_LENGTH / LEG_LENGTH) - 1) * SQRT(LEG_LENGTH * g)
Where g = 9.8 m/s^2 (gravitational constant)
Write a program to read in the data files from disk, it must then print the names of only the bipedal dinosaurs from fastest to slowest.
Do not print any other information.

```python
import csv

with open('dataset1.csv') as file1:
    rows = csv.DictReader(file1)
    for row in rows:
        print(row)
```

```python
import csv
bipedal = []
with open("dataset2.csv","r") as file2:
    rows = csv.DictReader(file2)
    for row in rows:
        if row["STANCE"] =='bipedal':
            bipedal.append(row)

print(bipedal)
```

```python
import csv

names = {}
for item in bipedal:
    names[item['NAME']]=item['STRIDE_LENGTH']
print(names)

```

```python
import math

final = []
with open("dataset1.csv","r") as file1:
    rows = csv.DictReader(file1)
    for row in rows:
        if row['NAME'] in names.keys():
            row['STRIDE_LENGTH']=names[row['NAME']]
            row['SPEED'] = ((float(names[row['NAME']])/float(row['LEG_LENGTH']))-1)*math.sqrt(float(row['LEG_LENGTH'])*9.8)

        if 'SPEED' in row and row["SPEED"]:
            final.append(row)

    final.sort(key=lambda x: float(x['SPEED']), reverse=True)

    for row in final:
        print(row['NAME'])
```
